# Modificación de Imágenes (OpenCV)

Técnicas fundamentales de transformación de imágenes utilizadas en visión por computadora y redes neuronales convolucionales (CNN).


#  1. Rotación

**Funciones:**
- `cv2.getRotationMatrix2D(center, angle, scale)`
- `cv2.warpAffine(img, M, dsize)`

**Parámetros:**
- `center`: Centro de rotación (x, y)
- `angle`: Ángulo en grados
- `scale`: Escala de la imagen
- `M`: Matriz de transformación
- `dsize`: Tamaño de salida


```python
import cv2
import numpy as np
import cv2

img = cv2.imread('IMG/30.jpeg')
img = cv2.resize(img, (400, 400))
(h, w) = img.shape[:2]#El valor devuelto por img.shape es una tupla con números enteros que indican el tamaño de cada dimensión
center = (w//2, h//2)#centrico de la imagen
M = cv2.getRotationMatrix2D(center, 34,0.8)#Matriz de rotación, el primer argumento es el centro de rotación, el segundo es el ángulo de rotación en grados, el tercero es la escala (1 significa sin cambio de tamaño) y el cuarto es un valor opcional que se utiliza para especificar la traslación después de la rotación (en este caso, no se aplica ninguna traslación).    
rotada = cv2.warpAffine(img, M, (w, h), flags=cv2.INTER_LINEAR, borderMode=cv2.BORDER_CONSTANT, borderValue=(0, 0, 0))
#La función cv2.warpAffine se utiliza para aplicar una transformación afín a una imagen. En este caso, se está aplicando la matriz de rotación M a la imagen img. El tercer argumento (w, h) especifica el tamaño de la imagen de salida después de la transformación. Los argumentos flags, borderMode y borderValue se utilizan para controlar cómo se manejan los bordes de la imagen durante la transformación. En este caso, se está utilizando la interpolación lineal (cv2.INTER_LINEAR) para calcular los valores de píxeles en la imagen de salida, y se está estableciendo el borde a un color blanco (255, 255, 255) utilizando el modo de borde constante (cv2.BORDER_CONSTANT).

cv2.imshow('figura',rotada)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

**Parámetros:**

#  4. Flip (Volteo)

**Función:** `cv2.flip()`

**Parámetros:**
- `0`: Vertical
- `1`: Horizontal
- `-1`: Ambos


```python
img = cv2.imread('IMG/30.jpeg')
img = cv2.resize(img, (300, 300))
Vertical=cv2.flip(img,  0)
Horizontal=cv2.flip(img,  1)
Ambos=cv2.flip(img, -1)
comparacion = np.hstack([img,Vertical,Horizontal,Ambos])
cv2.imshow('Original Vertical  Horizontal   Ambos', comparacion)
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

# 5. Conversión de Color

**Función:** `cv2.cvtColor()`

**Parámetros:**
- `img`: Imagen de entrada
- `code`: Tipo de conversión


```python
gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
cv2.imshow('GRIS', gris)
cv2.imshow('HSV', hsv)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# 6. Brillo y Contraste

**Función:** `cv2.convertScaleAbs()`

**Parámetros:**
- `alpha`: Contraste
- `beta`: Brillo
- # Contraste
- alpha: 0.0  → 3.0    `más allá de 3.0 satura demasiado`

- # Brillo  
- beta: -127  → 127    `límite natural de escala uint8` 

```python
ajuste = cv2.convertScaleAbs(img, alpha=1, beta=60)
comparacion = np.hstack([img, ajuste])
cv2.imshow('Original  |  Ajustado', comparacion)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
