```python
#con nombre de figura con contador de figuras sin roi
import cv2
import numpy as np

# ==============================
# ACTIVAR CAMARA
# ==============================
cap = cv2.VideoCapture(0)  # 0 = cámara principal

while True:
    ret, frame = cap.read()  # Captura un frame (imagen)
    if not ret:
        break

    # ==============================
    # CONVERTIR A HSV
    # ==============================
    # HSV facilita detectar colores
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # ==============================
    # RANGO DE COLOR (AMARILLO)
    # ==============================
    lower_yellow = np.array([20, 100, 100])
    upper_yellow = np.array([30, 255, 255])

    # ==============================
    # PARA ACTIVAR AZUL
    # ==============================
    lower_blue = np.array([100, 100, 100])
    upper_blue = np.array([140, 255, 255])

    # ==============================
    # PARA ACTIVAR ROJO
    # ==============================
    lower_red1 = np.array([0, 100, 100])
    upper_red1 = np.array([10, 255, 255])
    lower_red2 = np.array([160, 100, 100])
    upper_red2 = np.array([179, 255, 255])

    # ==============================
    # PARA ACTIVAR MORADO
    # ==============================
    lower_purple = np.array([130, 100, 100])
    upper_purple = np.array([160, 255, 255])

    # ==============================
    # CREAR MASCARA (FILTRAR COLOR)
    # ==============================
    mask_yellow = cv2.inRange(hsv, lower_yellow, upper_yellow)
    mask_blue   = cv2.inRange(hsv, lower_blue, upper_blue)
    mask_purple = cv2.inRange(hsv, lower_purple, upper_purple)
    mask_red1   = cv2.inRange(hsv, lower_red1, upper_red1)
    mask_red2   = cv2.inRange(hsv, lower_red2, upper_red2)

    # ==============================
    # LIMPIAR RUIDO (OPCIONAL)
    # ==============================
    kernel = np.ones((5,5), np.uint8)

    mask = mask_yellow + mask_blue + mask_purple + mask_red1 + mask_red2

    # ==============================
    # ENCONTRAR CONTORNOS
    # ==============================
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    contador = 0  # contador de figuras

    # ==============================
    # RECORRER CONTORNOS
    # ==============================
    for cnt in contours:
        area = cv2.contourArea(cnt)

        # FILTRAR OBJETOS PEQUEÑOS (ruido)
        if area > 500:
            contador += 1

            # ==============================
            # APROXIMAR FORMA
            # ==============================
            peri = cv2.arcLength(cnt, True)
            approx = cv2.approxPolyDP(cnt, 0.04 * peri, True)

            lados = len(approx)

            # ==============================
            # DETECTAR FIGURA
            # ==============================
            forma = "Desconocido"

            # TRIANGULOS
            if lados == 3:
                forma = "Triangulo"

            # CUADRADO / RECTANGULO
            elif lados == 4:
                x, y, w, h = cv2.boundingRect(approx)
                aspecto = float(w) / h

                if 0.9 <= aspecto <= 1.1:
                    forma = "Cuadrado"
                else:
                    forma = "Rectangulo"

            # PENTAGONO
            elif lados == 5:
                forma = "Pentagono"

            # HEXAGONO
            elif lados == 6:
                forma = "Hexagono"

            # MUCHOS LADOS (posibles estrellas o círculos)
            elif lados > 6:
                forma = "Figura compleja"

            # ==============================
            # INTENTOS BASICOS PARA FORMAS ESPECIALES
            # ==============================

            # ROMBO (similar a cuadrado pero inclinado)
            if lados == 4:
                x, y, w, h = cv2.boundingRect(approx)
                if abs(w - h) < 10:
                    forma = "Rombo"

            # ==============================
            # DIBUJAR CONTORNO
            # ==============================
            cv2.drawContours(frame, [cnt], -1, (0,255,255), 3)

            # ==============================
            # ENUMERAR FIGURAS + NOMBRE
            # ==============================
            x, y, w, h = cv2.boundingRect(cnt)

            cv2.putText(frame, f"{contador}: {forma}", (x, y-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,255), 2)

    # ==============================
    # MOSTRAR RESULTADOS
    # ==============================
    cv2.putText(frame, "Total: " + str(contador), (10, 40),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,255), 2)

    cv2.imshow("Camara", frame)
    cv2.imshow("Mascara", mask)

    # SALIR CON TECLA ESC
    if cv2.waitKey(1) & 0xFF == 27:
        break

# ==============================
# LIBERAR CAMARA
# ==============================
cap.release()
cv2.destroyAllWindows()
```
